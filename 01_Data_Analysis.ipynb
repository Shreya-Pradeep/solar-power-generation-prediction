```python
#create a new clean dataset which has the coumns that are used for prediction. Others have more missing values, so we can exclude them
df_model = df[['bat_current',
               'bat_status',
               'bat_volt',
               'load_current',
               'load_status',
               'load_volt',
               'pv_current',
               'pv_volt',
               'ft',
               'hour',
               'minute']].copy()
```

```python
df_model.head() #this is the new clean dataset
```

```python
#convert the words to numbers. machine learning models cannot understand the words. so instead we can convert it to numbers
df_model['bat_status'] = df_model['bat_status'].replace({
    'Charging': 0,
    'Discharging': 1,
    'Normal': 2,
    'Low': 3
})

df_model['ft'] = df_model['ft'].replace({
    'No Fault': 0,
    'Battery Low': 1
})
```

```python
df_model.head()
```

```python
df_model['load_status'] = df_model['load_status'].replace({
    'OFF': 0,
    'ON': 1
})
```

```python
df_model.head()
```

```python
df_model.to_csv("cleaned_solar_data.csv", index=False)
```
